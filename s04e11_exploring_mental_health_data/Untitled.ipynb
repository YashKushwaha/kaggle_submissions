```python
file_to_read = r'C:\Users\yash\Downloads\s04e11-mental-health-prediction-ensemble_2.py'
```

```python
with open(file_to_read, 'r') as f:
    text = f.read()
```

```python
lines = text.split('\n')
```

```python
lines_to_keep = []
for line in lines:
    if (line == '\n') or (line.startswith('#')) or (len(line) == 0) or {i for i in line} == {' '}:
        continue
    else:
        lines_to_keep.append(line)
```

```python
lines_to_keep[:5]
```

```python
final_text = '\n'.join(lines_to_keep)
```

```python
print(final_text)
```

```python
new_file_name = file_to_read.replace('.py' , '_cleaned.py')
```
