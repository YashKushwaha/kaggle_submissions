```python
def create_data_summary(df):
    isna_count = df.isna().sum()
    n_unique = pd.Series({col: df[col].nunique() for col in df.columns})
    dtypes = df.dtypes
    value_counts_top_5 = pd.Series({col: df[col].value_counts().head(5).to_dict() for col in df.columns})
    output = pd.DataFrame(dict(isna_count=isna_count, n_unique=n_unique, dtypes=dtypes, value_counts_top_5=value_counts_top_5))
    return output
```

```python
data_summary = create_data_summary(train_df)
```
